### 1. Creating Custom Functions

```python
# Lesson Workspace
beta_values = [0.86, 0.97, 0.81, 1.39, 1.05, 1.07]
print(max(beta_values))
print(min(beta_values))
```

```python
# Creating a custom function
# Run max and min together

def min_max(lst):
    """Run the python functions of max and min together on a list of values."""
    result =  f'The minimum value is: {min(lst)}\nThe maximum value is: {max(lst)}'
    return result

print(min_max(beta_values))
```

# Instructions
Write a python function that returns Future Value.  Use the values below:

formula = investment*(1 + growth_rate) ** years

data:
- investment = 100
- growth_rate = 0.1
- years = 3
- Future Value = 133.1

One way to round is to use the round built in function:
- round(number, decimal_places)


```python
# Solution to Exercise 1.1

def future_value(investment, growth_rate, years):
    """Calculates future growth of an investment."""
    result = round(investment * ( 1 + growth_rate ) ** years, 2)
    return f'The future value on a ${investment:,} investment with a {growth_rate:.2%} growth rate and held for {years} years will yield ${result:,}'

print(future_value(10_000, 0.1, 5))
```

# Instructions
1. Define a new function called calculate_ROI
    - The function should accept one argument named price_list
2. The function should return the result of the following formula
    - (target_price - current_price) / current_price
3. Create a list called companies that contains all the price lists defined in the first cell.
4. Write a for loop that performs the following actions:
    - Iterates through all the objects in companies
    - Applies the calculate_ROI function to each object
    - Rounds the result of each calculation to 3 decimal places
    - Prints the rounded result of each calculation.

```python
# Run this code cell to create the price lists below
# format of lists: [ current_price, target_price ]
jnj = [139.49, 144.20]
unh = [265.31, 284.14]
pfe = [42.96, 43.06]
abbv = [95.68, 107.44]
amgn = [200.58, 205.9]
abt = [68.42, 71.06]
```

```python
# Define a new function called calculate_roi

def calculate_roi(price_lst):
    """Given a list containing the current and target price, returns the roi."""
    return round((price_lst[1] - price_lst[0]) / price_lst[0],3)
```

```python
# Create a list of companies

companies = [jnj, unh, pfe, abbv, amgn, abt]

# for loop to run companies through calculate_roi
companies_roi = [calculate_roi(company) for company in companies]

print(companies_roi)
```
