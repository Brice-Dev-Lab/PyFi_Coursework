# Scenario:
Have a list of companies in a portfolio represented by different industries.  Wanted to know what industries were in the portfolio, but need to remove duplicates.  A changing a list to a set will remove the duplicates.


```python
# portfolio
portfolio = {
    'AAPL': 'IT',
    'AMZN': 'Consumer',
    'MSFT': 'IT',
    'GOOGL': 'IT',
    'BRK/A': 'Financials'
    }

# pull out the industries
industry_list = []
for industry in portfolio.values():
    industry_list.append(industry)
print(f'Industry list: ',industry_list)

# Change to a set
industry_set = set(industry_list)
print(f'Industry set: ', industry_set)
```

# Operations of Sets
- Use the format "*set_1.function(set_2)*"
- .union()
- .intersection()
- .difference()

```python
mkt_cap_over_500B = {'AAPL', 'AMZN', 'MSFT', 'GOOGL', 'BRK/A'}
aaa_rated = {'MSFT', 'JNJ', 'XOM'}
```

```python
# Combine sets with .union() - order doesn't matter
print(mkt_cap_over_500B.union(aaa_rated))
```

```python
# Find common elements with .intersection() - order doesn't matter
print(mkt_cap_over_500B.intersection(aaa_rated))
```

```python
# Find different elements with .difference() - order matters
# this looks at mkt_cap first to see what the unique values are in mkt_cap since it was listed first
print(mkt_cap_over_500B.difference(aaa_rated))
```

```python
# Find different elements with .difference() - order matters
# this looks at aaa_rated first to see what the unique values are in aaa_rated since it was listed first
print(aaa_rated.difference(mkt_cap_over_500B))
```
