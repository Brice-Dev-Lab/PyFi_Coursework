#### How to reshape the array
- use reshape method:
    - variable.reshape(num_rows, num_columns)
        - The num_rows and num_columns are the new shape desired
    - **numpy.reshape() does not modify the original array in-place — it returns a new reshaped array.**
        - Either print directly or save to another variable

- Transpose
- transpose: Swaps axes (like flipping rows & columns)
Think of it like rotating or flipping the array across its diagonal — it doesn't reshape arbitrarily, it specifically swaps axes.

```python
# reshape exercise 2.1 to be 2 rows and 3 columns
array_2_1 = array_2_1.reshape(3,2)
array_2_1_reshape = array_2_1.reshape(2, 3)
print(array_2_1_reshape)
```

## Instructions
- create a new array, array_b, and reshape it to 6 rows, 4 columns
- print array_b
- using ranges, print "new york stock exchange" making a new line every time you create a new word

```python
array_a = np.array([
    ['n', 'e', 'w', 'y', '_', 's'],
    ['t', 'o', 'c', 'k', '_', 'r'],
    ['_', '_', '_', 'k', 'e', 'x'],
    ['c', 'h', 'a', 'n', 'g', 'e']
])
```

```python
# Reshape to 6 rows and 4 columns and print
b = array_a.reshape(6, 4)
print(array_b)
```

```python
# Print 'new york stock exchange' from array_b (one word per line)
new = b[0:1, 0:3]
print(new)

york = b[0:4, 3:4]
york = york.reshape(1,4)
print(york)

sto = b[1:2, 1:]
ck = b[2:3, :2]
stock = np.concatenate([sto, ck], axis=1)  # horizontal
print(stock)

exch = b[4:5, 0:]
ange = b[5:6, 0:]
exchange = np.concatenate([exch, ange], axis=1)
print(exchange)
```

```python
current_prices = np.array([
    [223.84, 113.5, 139.49],
    [113.37, 30.37, 265.31],
    [1177.98, 54.73, 42.96],
])

target_prices = np.array([
    [232.27, 112.15, 144.2],
    [122.3, 34.33, 284.14],
    [1384.89, 62.12, 43.06],
])

companies = np.array([
    ['AAPL', 'JPM', 'JNJ'],
    ['MSFT', 'BAC', 'UNH'],
    ['FB', 'WFC', 'PFE']
])
```

```python
# Create price increases
price_increases = target_prices - current_prices
print(np.round(price_increases, 3))
```

```python
# Create returns
returns = price_increases / current_prices
print(np.round(returns, 3))
```

```python
# max value in return
max_return = np.max(returns)
print(f"Max Return: ", round(max_return, 3))

mean_return = np.mean(returns, axis=0)
print(f"mean by industry: ", np.round(mean_return, 3))
```

### 4. NumPy Random Tools

Working cells
- use of random function in numpy np.random.randint(low=1, high=11, size=10)
- if you want to hold the random numbers, then use variable.random.seed(random integers)
- To make random numbers unique, add *replace=False* to the argument


```python
np.random.seed(12)

random = np.random.randint(low=1, high=11, size=(5,2))
print(random)
```

```python
ordered_array = np.array(range(1, 11))
ordered_array
```

```python
np.random.seed(10)
print(np.random.choice(ordered_array, 5, replace=False))
```

```python
np.random.seed(99)  # For reproducibility

rand_array = np.array(range(1, 26))       # Ordered numbers 1 to 25
np.random.shuffle(rand_array)             # In-place shuffle
rand_array = rand_array.reshape(5, 5)     # Reshape into 5x5 matrix

print(rand_array)
```

```python
second_row = rand_array[1]  # Get second row (row index 1)

sample = np.random.choice(second_row, size=5, replace=True)
print("Random sample from the second row (with replacement):", sample)
```

```python
ordered_array_1 = np.array(range(1, 26))
ordered_array_1 = ordered_array_1.reshape(5,5)
print(ordered_array_1)
```

```python
third_column = ordered_array_1[:, 2]

sample_col = np.random.choice(third_column, size=5, replace=False)
print(sample_col)
```
