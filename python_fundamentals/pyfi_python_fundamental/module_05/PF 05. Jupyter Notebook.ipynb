### 3. New Object Type: Pandas Series

**Concepts:** A `Series` is a single labeled column in pandas. You can access columns from a DataFrame as Series, clean column names for consistent access, and create new derived columns using vectorized arithmetic across whole columns.

```python
# Practice
type(stock_data)
stock_data.Beta
```

```python
stock_data.columns = stock_data.columns.str.strip().str.replace(" ", "_").str.lower()
```

```python
print(stock_data.columns)
stock_data.current_price
```

<font color = 'blue'> **EXERCISE 3.1** </font>

![Instructions](figures/exercise_3_1_instructions.png)

**Concept check:** This exercise creates derived columns with vectorized math. `price_increase` measures target minus current price, and `return` converts that difference into a relative percentage-like ratio.

```python
stock_data.columns
```

```python
stock_data.head(1)
```

```python
stock_data['price_increase'] = round(stock_data.price_target - stock_data.current_price, 3)
print(stock_data.columns)
stock_data
```

```python
stock_data.head(1)
```

```python
stock_data['return'] = round(stock_data.price_increase / stock_data.current_price, 3)
stock_data
```

```python
print(stock_data.head(1))
stock_data['return'].describe()
```

### 4. Filtering Data with Boolean Masks

**Concepts:** A boolean mask is a True/False Series used to filter rows. Build masks from conditions (for example, sector equals a value), combine them with `&` and `|`, and use them to keep only records that match your criteria.

<font color = 'blue'> **EXERCISE 4.1** </font>

![Instructions](figures/exercise_4_1_instructions.png)

**Concept check:** Build a boolean mask from a condition (for example, `sector == 'Information Technology'`) and use it to filter rows. Converting the mask to `int` gives a 1/0 indicator column.

```python
stock_data.head(3)
```

```python
# create a boolean mask
boolean_mask = stock_data['sector'] == 'Information Technology'
stock_data[boolean_mask]
```

```python
# Indicator Variable for IT
stock_data['is_tech'] = (stock_data['sector'] == 'Information Technology').astype(int)
stock_data
```

### 5. Segmenting Data with Groupby

**Concepts:** `groupby` splits data into categories, applies an aggregation, and combines results. Here, grouping by `sector` and taking means summarizes typical numeric values per sector.

<font color = 'blue'> **EXERCISE 5.1** </font>

![Instructions](figures/exercise_5_1_instructions.png)

**Concept check:** `groupby('sector')` creates per-sector groups, then `.mean()` aggregates numeric columns so you can compare average metrics across sectors.

```python
stock_data.groupby('sector')[stock_data.select_dtypes(include='number').columns].mean()
```
