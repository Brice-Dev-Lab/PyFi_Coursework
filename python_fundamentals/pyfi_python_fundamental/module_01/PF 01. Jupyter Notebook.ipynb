# Exercise Instructions

- Calculate Enterpise Value and LTM EBITDA for both AAPL and MSFT and save as new variables
- Calculate the Enterprise Value / LTM EBITDA ratio for both AAPL and MSFT and save as new variables
    - Useful Formulae:

        - **Enterprise Value** = sum( Market Cap, Preferred Debt, LT Debt, Current LT Debt, Minority Interest ) - Cash & Equivalents

        - **LTM EBITDA** = sum( Q3 EBITDA('17), Q4 EBITDA('17), Q1 EBITDA('18), Q2 EBITDA('18) )

- Create a new variable that compares AAPL's ratio to MSFT's ratio using the **greater than** operator
- Use the **.format()** function to generate the following outputs:
    - AAPL EV = 940103
    - AAPL LTM EBITDA = 84070
    - AAPL EV / EBITDA = 11.18

  ---
    - MSFT EV = 699501
    - MSFT LTM EBITDA = 49468
    - MSFT EV / EBITDA = 14.14
  ---
    - AAPL's ratio is greater than MSFT's: False

```python
# Apple Data
a_mcap = 896473  # market cap
a_pref = 0  # preferred debt
a_ltdebt = 97128  # long term debt
a_cltdebt = 17472  # current long term debt
a_mint = 0  # minority interest
a_cash = 70970  # cash equivalence
a_q3_17 = 17067  # EBITDA
a_q4_17 = 30509  # EBITDA
a_q1_18 = 19699  # EBITDA
a_q2_18 = 16795  # EBITDA


# Microsoft Data
m_mcap = 757029  # market cap
m_pref = 0  # preferred debt
m_ltdebt = 72242  # long term debt
m_cltdebt = 3998  # current long term debt
m_mint = 0  # minority interest
m_cash = 133768  # cash equiv
m_q3_17 = 11155  # EBITDA
m_q4_17 = 12403  # EBITDA
m_q1_18 = 12042  # EBITDA
m_q2_18 = 13868  # EBITDA
 # EBITDA = Earnings Before Interest, Taxes, Depreciation, and Amortization
```

```python
# Calculate the Enterprise Value
# EV = Enterprise Value
aapl_ev = sum([a_mcap, a_pref, a_ltdebt, a_cltdebt, a_mint]) - a_cash
msft_ev = sum([m_mcap, m_pref, m_ltdebt, m_cltdebt, m_mint]) - m_cash

# Calculate the LTM EBITDA
aapl_ltm_ebitda = sum([a_q3_17, a_q4_17, a_q1_18, a_q2_18])
msft_ltm_ebitda = sum([m_q3_17, m_q4_17, m_q1_18, m_q2_18])

# Calculate the ratio = EV / EBITDA ratio
aapl_ratio = aapl_ev / aapl_ltm_ebitda
msft_ratio = msft_ev / msft_ltm_ebitda

# Generate Report
print("AAPL EV = {} ".format(aapl_ev))
print("AAPL LTM EBITDA = {}".format(aapl_ltm_ebitda))
print("AAPL EV / EBITDA = {:.2f}".format(aapl_ratio))
print("---")
print("MSFT EV = {}".format(msft_ev))
print("MSFT LTM EBITDA = {}".format(msft_ltm_ebitda))
print("MSFT EV / EBITDA = {:.2f}".format(msft_ratio))
print("---")
print("AAPL's ratio is greater than MSFT's ratio: {}".format(aapl_ratio > msft_ratio))
```
