### 3. New Object Type: Pandas Series

```python
# Display first row of data
print(stock_data.head(1))

# Create new features
stock_data['Price Increase'] = stock_data['Price Target'] - stock_data['Current Price']
stock_data['Return'] = stock_data['Price Increase'] / stock_data['Current Price']
stock_data.head(1)
```

### 4. Filtering Data with Boolean Masks

```python
# display first 3 rows of stock data
stock_data.head(3)
```

```python
# create a boolean mask for Information Technology companies

is_info_tech = (stock_data['Sector'] == 'Information Technology').astype(int)
stock_data['is_IT'] = is_info_tech
stock_data.head(3)
```
